## 3.1 Введение

`Altair` &ndash; это *декларативная* библиотека для визуализации данных на `Python`, основанная на грамматике `Vega` и `Vega-Lite`. Библиотека реализует концепцию ***«грамматики графики»*** (*The Grammar of Graphics*). Данная математическая и теоретическая концепция описывает любой статистический график как структурированную  комбинацию независимых модульных слоев. Вместо того чтобы воспринимать графики как фиксированные шаблоны («круговая диаграмма», «точечный график»), грамматика предлагает собирать их из базовых элементов. Эту концепцию сформулировал Леланд Уилкинсон в 1999 году, а всемирную популярность она обрела благодаря Хэдли Уикхему и его знаменитой библиотеке `ggplot2` для языка программирования `R`.  

В современной реализации (например, в `ggplot2`) любой график состоит из следующих уровней, накладывающихся друг на друга:

Преимущества использования `Altair` для анализа и визуализации данных:

  Одним из ключевых преимуществ `Altair` является его декларативный синтаксис. Это существенно упрощает процесс создания визуализаций и снижает вероятность ошибок в коде.

  `Altair` предоставляет множество инструментов для добавления интерактивности к вашим визуализациям. Вы можете легко создавать всплывающие подсказки, устанавливать возможность масштабирования и перемещения графиков, а также добавлять инструменты для фильтрации данных в реальном времени. Это особенно полезно при создании дашбордов и интерактивных отчетов.
- Интеграция с библиотеками для анализа данных  
  `Altair` прекрасно интегрируется с другими популярными библиотеками для анализа данных в Python, такими как `pandas`. Это означает, что вы можете легко использовать свои существующие навыки работы с данными в сочетании с мощными средствами визуализации `Altair`.
- Компактный и читаемый код  
  Генерируемый `Altair` код обычно более компактный и читаемый, чем код для создания графиков в других библиотеках. Это делает его отличным выбором для тех, кто стремится к ясности и понятности кода.
- Открытость и активное сообщество  
  `Altair` &ndash; это open-source проект с активным сообществом разработчиков и пользователей.
- Красивые и информативные графики  
  `Altair` уделяет большое внимание деталям и возможностям кастомизации графиков. Вы можете легко настраивать цвета, шрифты, размеры и стили, чтобы создать визуализации, которые соответствуют вашим требованиям и корпоративному стилю.
- Поддержка разных форматов данных  
  `Altair` позволяет работать с данными в разных форматах, включая CSV, Excel, JSON и даже базы данных. Это делает его мощным инструментом для анализа данных, не зависящим от источника данных.

## 3.2 Базовые концепции `Altair`

Грамматика визуализации данных в Altair &ndash; это декларативный подход к построению графиков, основанный на спецификациях `Vega` и `Vega-Lite`, где пользователь описывает ***что*** он хочет увидеть на графике, а не ***как*** его пошагово рисовать. Вместо ручного конфигурирования осей и цветов, вы связываете («кодируете») переменные из вашего датасета с визуальными свойствами геометрических объектов.

В `Altair` грамматика графики выражается через простую формулу: `Данные` + `Геометрические маркеры (Marks)` + `Визуальное кодирование (Encodings)`.

Типы кодировок:
 - `Q` (quantitative &ndash; количественные)
 - `N` (nominal &ndash; неупорядоченные)
 - `O` (ordinal &ndash; упорядоченные)
 - `T` (temporal &ndash; временные)

### Базовый график

Ключевые элементы графика Altair &ndash; это данные (**data**), метки (**marks**) и кодировки (**encodings**).

Их задание в коде обычно включает:
- создание объекта графика с данными
- выбор типа меток
- добавление кодировок, каждая из которых сопоставляет столбец данных визуальному каналу (ось, цвет, размер и т. п.)

Пример:
```python
alt.Chart(data).mark_point().encode(
    encoding_1='column_1',
    encoding_2='column_2',
    # и т.д.
)
```

Рассмотрим эти элементы по очереди.

#### Данные

В `Altair` данные базируются на структуре `pandas.DataFrame`. Будем использовать набор данных о машинах (`cars`); его можно загрузить через пакет `vega_datasets`.

````python
import altair as alt
from vega_datasets import data

alt.__version__
````

````python
cars = data.cars()
cars.head()
````

Данные в `Altair` ожидаются в аккуратной форме (**tidy**), то есть:
- каждая строка &ndash; это одно наблюдение
- каждый столбец &ndash; это отдельная переменная

#### Объект `Chart`

Определив данные, можно создать фундаментальный объект `Altair` &ndash; `Chart`. По сути, `Chart` &ndash; это объект, формирующий JSON-словарь, состоящий из данных и визуальных кодировок, который можно передать для отрисовки библиотекой `Vega-Lite`.

Для наглядности можно посмотреть, как выглядит это JSON-представление для одной строки данных. На этом этапе график включает JSON-представление `DataFrame`, тип отметки и некоторую метаинформацию.

````python
cars1 = cars.iloc[:1]
alt.Chart(cars1).mark_point().to_dict()
````

#### Метки (Marks)

Можно выбрать вид меток, который будет использоваться для представления данных. В предыдущем примере был выбран тип меток `point`, отображающий каждое наблюдение в виде точки.

````python
alt.Chart(cars).mark_point()
````

Результат &ndash; визуализация с одной точкой на каждую строку данных, хотя пока это не слишком информативно: все точки наложены друг на друга.

````python
alt.Chart(cars1).mark_point().to_dict()
````

Некоторые распространённые типы меток:
- `mark_point()`
- `mark_circle()`
- `mark_square()`
- `mark_line()`
- `mark_area()`
- `mark_bar()`
- `mark_tick()`

#### Кодировки (Encodings)

Следующий шаг &ndash; добавить визуальные каналы кодировок (*encodings*) к графику. Канал кодировки задаёт, как конкретный столбец данных сопоставляется визуальному свойству на графике.

Наиболее часто используемые каналы:
- `x`: значение по оси X
- `y`: значение по оси Y
- `color`: цвет метки
- `opacity`: прозрачность метки
- `shape`: форма метки
- `size`: размер метки
- `row`: строка в сетке 
- `column`: столбец в сетке

Полный список см. в документации: [Vega-Altair &rarr; Encodings](https://altair-viz.github.io/user_guide/encodings/index.html).

Кодировки задаются методом `encode()` объекта `Chart`. Начнём с отображения `Origin` по оси `y`.

````python
alt.Chart(cars).mark_point().encode(
    y='Origin'
)
````

В JSON-спецификации при этом появляется ключ `encoding`:

````python
alt.Chart(cars1).mark_point().encode(
    y='Origin'
).to_dict()
````

Добавим `Miles_per_Gallon` по оси `x`:

````python
alt.Chart(cars).mark_point().encode(
    y='Horsepower',
    x='Miles_per_Gallon',
    row = 'Cylinders:O',
    column = 'Origin:N'
)
````

Или, например:

````python
alt.Chart(cars).mark_point().encode(
    color='Origin',
    y='Miles_per_Gallon',
    x='Year'
)
````

#### Типы кодировок (Encoding Types)

Одна из ключевых идей Altair &ndash; библиотека стремится автоматически выбирать подходящие шкалы по умолчанию для типов данных.

Базовые типы:
- `quantitative (Q)`: числовые величины (вещественные)
- `nominal (N)`: имена / неупорядоченные категориальные
- `ordinal (O)`: упорядоченные категориальные
- `temporal (T)`: дата/время

Если вы передаёте данные как `pandas.DataFrame`, типы будут определены автоматически. Если вы указываете данные по URL, необходимо явно задавать типы для каждого столбца.

Рассмотрим простой график с тремя столбцами из набора `cars`:

````python
alt.Chart(cars).mark_tick().encode(
    x='Miles_per_Gallon',
    y='Origin',
    color='Cylinders'
)
````

Зададимся вопросами:
- какой тип лучше подходит для `Miles_per_Gallon`?
- какой тип лучше подходит для `Origin`?
- какой тип лучше подходит для `Cylinders`?

Зададим типы данных, используя односимвольные коды (например, заменим `"Miles_per_Gallon"` на `"Miles_per_Gallon:Q"`, чтобы явно указать количественный тип).

````python
alt.Chart(cars).mark_tick().encode(
    x='Miles_per_Gallon:Q',
    y='Origin:N',
    color='Cylinders:O'
)
````

Обратим внимание, как изменился график, когда для `'Cylinders'` указали порядковый тип (`ordinal`).
> На практике полезно всегда явно задавать типы, особенно при работе с данными из файла или по URL, где это обязательно.

## 3.3 Биннинг и агрегация данных

Мы уже обсудили данные (`data`), отметки (`marks`), кодировки (`encodings`) и типы кодировок. Следующий важный элемент API `Altair` &ndash; подход к биннингу и агрегированию данных.

### Group-By в Pandas

В `Altair` шаблон *«split-apply-combine»* можно задать, передав оператор агрегации в строковой форме внутри кодировки. Например, диаграмму со средними значениями по группам можно получить, указав `x='mean(Miles_per_Gallon)'` при группировке по `y='Origin'`. Обратите внимание, что группировка происходит неявно внутри кодировок: здесь мы группируем только по `Origin`, затем считаем среднее в каждой группе.

````python
cars.groupby('Origin')['Miles_per_Gallon'].mean()
````

````python
alt.Chart(cars).mark_bar().encode(
    y='Origin',
    x='mean(Miles_per_Gallon)'
)
````

### Одномерный биннинг: гистограммы

Одна из самых распространённых задач биннинга &ndash; построение гистограмм. Например, можно построить гистограмму по `Miles_per_Gallon`, задав для оси `x` биннинг (`bin=True`) и для `y` &ndash; подсчёт (`count()`). Цвет (`color`) можно сопоставить с `Origin`.

````python
alt.Chart(cars).mark_bar().encode(
    x = alt.X('Miles_per_Gallon', bin=True),
    y = alt.Y('count()'),
    color = alt.Color('Origin')
)
````

Декларативный подход `Altair` позволяет легко переопределять назначение закодированных величин между каналами и получать разные представления одних и тех же данных. Например, если мы группируем интервалы по `Miles_per_Gallon` и назначаем эти значения каналу `color`, а по оси `x` отображаем `count()`, по оси `y` &ndash; `Origin`, то получим иной вид распределения.

````python
alt.Chart(cars).mark_bar().encode(
    color=alt.Color('Miles_per_Gallon', bin=True),
    x='count()',
    y='Origin'
)
````

Такой подход помогает лучше оценивать пропорции `MPG` внутри каждой страны. При желании можно нормировать количество по оси `x`, чтобы сравнивать доли напрямую, задав параметр `stack='normalize'` у канала `x`.

````python
alt.Chart(cars).mark_bar().encode(
    color=alt.Color('Miles_per_Gallon', bin=True),
    x=alt.X('count()', stack='normalize'),
    y='Origin'
)
````

Из такого сравнения видно, что существенно более половины автомобилей из США попали в «низкую экономичность». Если снова изменить кодировку и, например, отобразить `Miles_per_Gallon` (с биннингом через `maxbins`) по оси `x`, `Origin` по оси `y`, а `color` сопоставить с `count()`, то получим тепловую карту (*heatmap*) того же набора данных.

````python
alt.Chart(cars).mark_rect().encode(
    x=alt.X('Miles_per_Gallon', bin=alt.Bin(maxbins=20)),
    color='count()',
    y='Origin',
)
````

Этот пример иллюстрирует важную особенность грамматики `Altair`: разные типы графиков (например, 2D-heatmap и составная столбчатая гистограмма) могут кодировать одну и ту же информацию, просто иным способом.

### Другие функции агрегации

Агрегаторы применимы и к данным, которые «неявно» бьются на группы. Например, если отобразить `Miles_per_Gallon` по оси `y` во времени (`x='Year:T'`) и задать `color='Origin'`, точки часто будут перекрываться, и важные особенности станут менее заметны.

````python
alt.Chart(cars).mark_point().encode(
    x='Year:T',
    color='Origin',
    y='Miles_per_Gallon'
)
````

Сделать картину нагляднее можно, построив среднее в каждой группе (здесь &ndash; в разрезе комбинаций «год/страна»), заменив точки на линии со значениями `y='mean(Miles_per_Gallon)'`.

````python
alt.Chart(cars).mark_line().encode(
    x='Year:T',
    color='Origin',
    y='mean(Miles_per_Gallon)'
)
````

Однако один `mean` показывает лишь часть картины. В `Altair` есть встроенные инструменты для вычисления нижней и верхней границ доверительного интервала для среднего. Можно использовать `mark_area()` и задать нижнюю границу через `y='ci0(Miles_per_Gallon)'`, а верхнюю &ndash; через `y2='ci1(Miles_per_Gallon)'`, сохраняя `x='Year:T'` и `color='Origin'`. Полупрозрачная заливка (через `opacity`) поможет визуально передать доверительный интервал. Подробнее: [Vega-Altair &rarr; Distribution Statistics](https://altair-viz.github.io/user_guide/transform/aggregate.html#distribution-statistics).

````python
alt.Chart(cars).mark_area(opacity=0.3).encode(
    x='Year:T',
    color='Origin',
    y='ci0(Miles_per_Gallon)',
    y2='ci1(Miles_per_Gallon)'
)
````

### Группировка по времени

Особый случай биннинга &ndash; группировка временных значений по компонентам даты: например, по месяцу года или дню месяца. Рассмотрим простой набор данных со среднесуточными температурами в Сиэтле.

````python
temps = data.seattle_temps()
temps.head()
````

Если попытаться отобразить все исходные строки напрямую, `Altair` выдаст `MaxRowsError`: по умолчанию работа с наборами более чем в 5000 строк, чтобы избежать чрезмерно тяжёлых файлов и уменьшения производительности.

````python
alt.Chart(temps).mark_line().encode(
    x='date:T',
    y='temp:Q'
)
````

````python
len(temps)
````

### Немного дополнительной информации: как `Altair` кодирует данные

Ограничение в 5000 строк введено на основе наблюдений за практикой: если не задумываться о способе представления данных, легко получить очень большой файл с низкой производительностью. Когда вы передаёте `pandas.DataFrame` в график `Altair`, данные конвертируются в JSON и сохраняются в спецификации графика. Эта спецификация встраивается в выход вашего ноутбука; несколько десятков таких графиков с крупными наборами данных могут существенно замедлить работу машины.

Как обойти `MaxRowsError`:
1) Использовать меньший набор данных. Например, агрегировать температуры по дате (усреднить по дням) с помощью `pandas` и затем отобразить результат.
2) Отключить проверку `MaxRowsError`, задав в трансформерах данных режим без ограничения. Следует помнить о риске получить очень большой файл.
3) Раздавать данные с локального потокового сервера; пакет `altair_data_server` упрощает настройку. Этот подход может не работать в некоторых облачных сервисах `Jupyter`.
4) Использовать `URL`, указывающий на источник данных (например, хранить данные в gist на `gist.github.com`). В `vega_datasets` у источников есть свойство `url`, которое можно напрямую передать в Altair &ndash; тогда в спецификацию попадёт только ссылка, а не весь массив данных.

В данном примере удобно и производительно воспользоваться последним вариантом: использовать `url` из `vega_datasets`, так что в спецификацию будет включён только адрес источника, а не весь набор.

````python
temps = data.seattle_temps.url
````

````python
alt.Chart(temps).mark_line().to_dict()
````

После этого снова построим график (`x='date:T'`, `y='temp:Q'`).

````python
alt.Chart(temps).mark_line().encode(
    x='date:T',
    y='temp:Q'
)
````

Полученный график выглядит перегруженным. Предположим, мы хотим разделить данные по месяцам. Это делается посредством `TimeUnit Transform` для даты: канал `x` можно задать как `month(date)`.

````python
alt.Chart(temps).mark_point().encode(
    x=alt.X('month(date):T'),
    y='temp:Q'
)
````

График станет ещё нагляднее, если затем агрегировать температуры &ndash; например, через `y='mean(temp):Q'` и тип `x` сделать порядковым (`'month(date):O'`), чтобы месяцы располагались в естественном порядке.

````python
alt.Chart(temps).mark_bar().encode(
    x=alt.X('month(date):O'),
    y='mean(temp):Q'
)
````

Можно также разделять даты двумя способами и получать интересные виды данных. Например:
- тепловая карта, где по оси `x` &ndash; `date(date):O` (день месяца), по `y` &ndash; `month(date):O` (месяц), а цвет &ndash; `mean(temp):Q`

````python
alt.Chart(temps).mark_rect().encode(
    x=alt.X('date(date):O'),
    y=alt.Y('month(date):O'),
    color='mean(temp):Q'
)
````

- или средняя почасовая температура как функция месяца: по `x` &ndash; `hours(date):O`, по `y` &ndash; `month(date):O`, цвет &ndash; `mean(temp):Q`.

````python
alt.Chart(temps).mark_rect().encode(
    x=alt.X('hours(date):O'),
    y=alt.Y('month(date):O'),
    color='mean(temp):Q'
)
````

Такие преобразования особенно полезны при работе с временными данными. Подробнее о `TimeUnit Transform` см. в [Vega-Altair &rarr; TimeUnit](https://altair-viz.github.io/user_guide/transform/timeunit.html).

## 3.4 Составные графики

`Altair` предоставляет лаконичное API для создания многопанельных и многослойных графиков. Ниже кратко рассмотрены четыре подхода:

- Слои (Layering)
- Горизонтальная конкатенация (Horizontal Concatenation)
- Вертикальная конкатенация (Vertical Concatenation)
- Повторяющиеся сетки графиков (Repeat Charts)

Далее &ndash; кратко рассмотрим каждый из них.

### Слои (Layering)

Слои позволяют размещать несколько отметок на одном объекте `Chart`. Типичный пример &ndash; объединение точек и линий, представляющих одни и те же данные.

В качестве примера используется набор данных `stocks` (цены акций во времени). Сначала можно построить простой линейный график (`mark_line`) по полям `date:T`, `price:Q` и `symbol:N`. Аналогично строится точечный график (`mark_circle`) с теми же кодировками.

````python
stocks = data.stocks()
stocks.head()
````

````python
alt.Chart(stocks).mark_line().encode(
    x='date:T',
    y='price:Q',
    color='symbol:N'
)
````

````python
alt.Chart(stocks).mark_circle().encode(
    x='date:T',
    y='price:Q',
    color='symbol:N'
)
````

Эти два представления можно объединить с помощью оператора `+`, получив единый слой, в котором одновременно присутствуют линии и точки. Оператор `+` является сокращением для функции `alt.layer()`, которая выполняет ту же задачу.

````python
lines = alt.Chart(stocks).mark_line().encode(
    x='date:T',
    y='price:Q',
    color='symbol:N'
)

points = alt.Chart(stocks).mark_circle().encode(
    x='date:T',
    y='price:Q',
    color='symbol:N'
)

lines + points
````

````python
alt.layer(lines, points)
````

Удобный приём &ndash; создать «базовый» график (`base`) с общими кодировками (например, `x='date:T'`, `y='price:Q'`, `color='symbol:N'`), а затем комбинировать производные варианты одной командой (например, `base.mark_line() + base.mark_circle()`).

````python
base = alt.Chart(stocks).encode(
    x='date:T',
    y='price:Q',
    color='symbol:N'
)

base.mark_line() + base.mark_circle()
````

### Горизонтальная конкатенация (Horizontal Concatenation)

Помимо наслаивания, можно располагать графики рядом по горизонтали. Это достигается с помощью функции `alt.hconcat(...)` или эквивалентного оператора `|`.

````python
base.mark_line() | base.mark_circle()
````

````python
alt.hconcat(base.mark_line(),
            base.mark_circle())
````

Такой подход полезен для создания многопанельных представлений (*multi-panel views*). Например, для набора `iris` можно определить базовую точечную диаграмму (`mark_point`) с `x='petalWidth'`, `y='petalLength'`, `color='species'`, а затем расположить рядом её вариант с изменённой осью `x` (например, `x='sepalWidth'`), используя оператор `|`.

````python
iris = data.iris()
iris.head()
````

````python
base = alt.Chart(iris).mark_point().encode(
    x='petalWidth',
    y='petalLength',
    color='species'
)

base | base.encode(x='sepalWidth')
````

### Вертикальная конкатенация (Vertical Concatenation)

Вертикальная конкатенация по смыслу аналогична горизонтальной, но графики располагаются один под другим. Для этого используется оператор `&` или соответствующая функция (`alt.vconcat`, если применять явный вызов).

Продолжая пример с `iris`, расположим исходный базовый график и его вариант с `y='sepalWidth'` вертикально: `base & base.encode(y='sepalWidth')`.

````python
base & base.encode(y='sepalWidth')
````

### Повторяющиеся сетки графиков (Repeat Chart)

Поскольку часто требуется горизонтально и вертикально конкатенировать графики, меняя одну из кодировок (например, ось `x` или `y`), в `Altair` предусмотрен удобный механизм `repeat()`.

Шаблон таков:
- определить список полей (например, `['petalLength', 'petalWidth', 'sepalLength', 'sepalWidth']`)
- в кодировках осей использовать `alt.repeat("column")` и `alt.repeat("row")` с указанием типа (обычно `quantitative` для числовых признаков)
- задать размеры отдельных панелей через `properties(width=..., height=...)`
- вызвать `.repeat(row=..., column=...)`, передав списки полей для строк и столбцов
- опционально добавить `.interactive()` для включения интерактивности

Так создаётся матрица рассеяния, в которой каждая панель &ndash; это комбинация пары признаков, а цвет, например, кодирует вид (в случае `iris` &ndash; `species`).

Синтаксис `repeat` может показаться менее удобным, чем хотелось бы, однако он позволяет существенно сократить шаблонный код для многомерных раскладок.

````python
fields = ['petalLength', 'petalWidth', 'sepalLength', 'sepalWidth']

alt.Chart(iris).mark_point().encode(
    alt.X(alt.repeat("column"), type='quantitative'),
    alt.Y(alt.repeat("row"), type='quantitative'),
    color='species'
).properties(
    width=200,
    height=200
).repeat(
    row=fields,
    column=fields[::-1]
).interactive()
````
